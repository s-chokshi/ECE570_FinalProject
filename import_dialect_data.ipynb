```python
import json

with open('sentences_ch_de_numerics.json') as f:
    data = json.load(f)

with open('ch_sg.txt', 'w') as f_sg, open('de.txt', 'w') as f_de:
    for entry in data:
        if 'ch_sg' in entry and 'de' in entry:
            f_sg.write(entry['ch_sg'] + '\n')
            f_de.write(entry['de'] + '\n')
```
